```python
import pandas as pd
```

```python
data = pd.read_csv("C:/Users/Shubham Baraiya/Downloads/Palo Alto Networks.csv")
print(data)
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
df.isnull().sum()
```

```python
df['Engagement_Score'] = (df['EnvironmentSatisfaction']+df['JobInvolvement']+df['JobSatisfaction']+df['RelationshipSatisfaction'])/4
```

```python
print(df['Engagement_Score'])
```

```python
def burnout_risk(row):
    if row['OverTime']=='Yes' and row['WorkLifeBalance']<=2:
        return 'High'
    elif row['OverTime']=='Yes' or row['WorkLifeBalance']<=2:
        return 'Medium'
    else:
        return 'Low'
        
df['Burnout_Risk_Id'] = df.apply(burnout_risk,axis=1)
```

```python
print(df['Burnout_Risk_Id'])
```

```python
print(df['Burnout_Risk_Id'].value_counts())
```

```python
df.groupby('OverTime')['Engagement_Score'].mean()
```

```python
df.groupby('BusinessTravel')['Engagement_Score'].mean()
```

```python
def Category(x):
    if x<=15:
        return 'Short'

    else:
        return 'Long'
df['distance_Category']=df['DistanceFromHome'].apply(Category)
```

```python
df['distance_Category'].value_counts()
```

```python
df.groupby('distance_Category')['Engagement_Score'].mean()
```

```python
df.groupby('JobLevel')['Engagement_Score'].mean()

```

```python
df.groupby(pd.cut(df['YearsAtCompany'], bins=[0,5,10,20,40]))['Engagement_Score'].mean()
```

```python
df.groupby(pd.cut(df['YearsInCurrentRole'],bins=[0,5,10,15,20]))['Engagement_Score'].mean()
```

```python
df['Attrition_Level'] = df['Attrition'].map({0:'Stay',1:'Left'})
```

```python
df.groupby('Attrition_Level')['Engagement_Score'].mean()
```

```python
df.to_csv("final_data.csv", index=False)
```

```python
df['OverTime_Flag'] = df['OverTime'].map({'Yes':1,'No':0})
df['Low_WLB'] = df['WorkLifeBalance'].apply(lambda x: 1 if x<=2 else 0)
df['Burnout_Score'] = df['OverTime_Flag']+df['Low_WLB']
print(df['Burnout_Score'])
```

```python
df.to_csv("final_data.csv", index=False)
```

```python
df['Satisfaction_Availability_Score'] = df[['EnvironmentSatisfaction','JobSatisfaction','RelationshipSatisfaction']].std(axis=1)
```

```python
print(df['Satisfaction_Availability_Score'])
```

```python
df['Travaler_Num'] = df['BusinessTravel'].map({'Travel_Frequently':1,'Travel_Rarely':0,'Non-Travel':0})
```

```python
df['Workload_Stress_Indicator'] = df['OverTime_Flag']+df['Travaler_Num']
```

```python
print(df['Workload_Stress_Indicator'])
```

```python
df.to_csv("final_data.csv", index=False)
```
